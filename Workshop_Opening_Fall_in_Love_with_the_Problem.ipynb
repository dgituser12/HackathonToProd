# Fall in Love with the Problem, Not the Solution

### From technology-first thinking to outcome-first engineering

> **The problem is the anchor. The solution is allowed to evolve.**

# We Are Surrounded by Solutions

Today, new technologies arrive constantly:

| Models | RAG | Agents | APIs |
|---|---|---|---|
| Computer Vision | Vector Databases | Cloud | Automation |

It is natural to ask:

> **“What can I build with this?”**

That is a useful question.

But it is not always the best question to ask first.

# Start Somewhere Else

# What problem are we solving?

Before choosing a technology, ask:

- **Who has the problem?**
- **How painful is it?**
- **How is it solved today?**
- **What would better look like?**

Only then ask:

> **What technology might help?**

# Fall in Love with the Problem

# Fall in love with the problem, not the solution.

If you fall in love with your **solution**, you start defending it.

If you fall in love with the **problem**, you keep looking for a better way to solve it.

A startup may begin with a chatbot, discover that users really need search, and later discover that workflow automation is better.

That is not failure.

**That is learning.**

# The Solution Is Allowed to Change

### Solution-first thinking

```text
“I built an AI chatbot.”
          ↓
Find something for it to do
          ↓
Defend the chatbot
```

### Problem-first thinking

```text
“There is a real problem.”
          ↓
Understand it deeply
          ↓
Try a solution
          ↓
Measure what happens
          ↓
Change the solution
          ↓
Learn again
```

**The problem remains the anchor.**

# This Is an Engineering Mindset

| Solution-first | Problem-first |
|---|---|
| Which model should I use? | What are we trying to improve? |
| How do I make the prompt better? | Why is the current process failing? |
| Can we automate this? | Should we automate this? |
| Can we build it? | Should we build it? |
| Is the demo impressive? | Does it create value? |

> **This is not only a startup mindset. It is an engineering mindset.**

# A Quick Thought Experiment

## HelpDeskAI

Imagine someone tells you:

> “Our support team receives thousands of customer tickets every month. Agents spend time manually routing them to the appropriate team.”

You are asked to build an AI solution.

### What would you want to know before writing a single line of code?

Take a moment.

Ask questions.

Do not design the solution yet.

# You Just Started Engineering

Good questions might include:

- How many tickets arrive each month?
- How long does manual routing take?
- Who performs the routing?
- What categories exist?
- What happens when a ticket is routed incorrectly?
- How costly is a mistake?
- What does success look like?
- What happens today?
- What information would the system need?

None of these questions require an AI model.

> **Understanding the problem is engineering work.**

```text
┌──────────────────┐
│      PROBLEM     │
│ What hurts?      │
└────────┬─────────┘
         ↓
┌──────────────────┐
│      OUTCOME     │
│ What should      │
│ improve?         │
└────────┬─────────┘
         ↓
┌──────────────────┐
│     SOLUTION     │
│ What might help? │
└────────┬─────────┘
         ↓
┌──────────────────┐
│   ENGINEERING    │
│ Can we build it, │
│ measure it, and  │
│ operate it?      │
└──────────────────┘
```

**Technology is a means, not the destination.**

**The problem remains the anchor.**
